## Data Preprocessing

### Missing Value Analysis

The dataset was checked for missing values to ensure data quality and identify any null entries before performing further analysis.

```python
df.isnull().sum()
```

### Handling Missing Values

Missing values in the review rating column were imputed using the median rating of each product category to maintain data consistency and reduce the impact of outliers.

```python
df["Review Rating"] = df.groupby("Category")["Review Rating"].transform(lambda x: x.fillna(x.median()))
```

### Verifying Missing Values

The dataset was re-evaluated after data preprocessing to confirm that missing values were successfully handled.

```python
df.isnull().sum()
```

### Standardizing Column Names

Column names were transformed into a consistent format by converting them to lowercase, replacing spaces with underscores, and renaming columns for better readability and easier querying.

```python
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ','_')
df = df.rename(columns = {"purchase_amount_(usd)" : "purchase_amount"})
```

### Viewing Updated Column Names

The updated column names were displayed to verify successful standardization and ensure consistency across the dataset.

```python
df.columns
```

## Feature Engineering

### Creating Age Groups

A new age group feature was created by dividing customers into quartile-based categories to support demographic analysis and customer segmentation.

```python
# create a column age_group
labels = ["Young Adult", "Adult", "Middle-aged", "Senior"]
df["age_group"] = pd.qcut(df["age"], q = 4, labels = labels)
```

### Previewing Age Group Segmentation

The newly created age group categories were displayed alongside customer ages to validate the segmentation process.

```python
df[["age","age_group"]].head(100)
```

### Converting Purchase Frequency into Numerical Format

Purchase frequency values were transformed into numerical day-based intervals to enable quantitative analysis and customer purchasing behavior evaluation.

```python
# create column purchase_frequency_days

frequency_mapping = {
    "Fortnightly": 14,
    "Weekly": 7,
    "Monthly": 30,
    "Quarterly": 90,
    "Bi-Weekly": 14,
    "Annually": 365,
    "Every 3 Months": 90
}

df["purchase_frequency_days"] = df["frequency_of_purchases"].map(frequency_mapping)
```

### Validating Purchase Frequency Conversion

The original purchase frequency values and their corresponding numerical representations were displayed to verify the accuracy of the transformation process.

```python
df[["frequency_of_purchases", "purchase_frequency_days"]].head(10)
```

### Reviewing Discount and Promotion Data

Discount application and promotional code usage data were examined to understand customer engagement with promotional offers.

```python
df[["discount_applied","promo_code_used"]].head(10)
```

### Checking Data Consistency

A validation check was performed to verify consistency between discount application and promotional code usage records.

```python
(df["discount_applied"] == df["promo_code_used"]).all()
```

### Removing Redundant Features

The promotional code column was removed from the dataset after confirming redundancy to improve data clarity and reduce unnecessary attributes.

```python
df = df.drop("promo_code_used", axis = 1)
```
